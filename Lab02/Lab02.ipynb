```python
## Task 06 ##
regions = np.array(["Punjab", "Sindh", "KP", "Balochistan", "Islamabad", "AJK"])
p17 = np.array([110.01, 47.85, 35.53, 12.34, 2.01, 4.05])
census23 = np.array([127.69, 55.70, 40.85, 14.89, 2.36, 4.05])

def pairs_dict(dice_pairs):
    out = {}
    for kar, v in dice_pairs:
        out[str(kar)] = v.item() if hasattr(v, "item") else v
    return out

total = census23.sum()
print(f"Task06a : total population 2023 = {total:.2f} million")

pct_share = (census23 / total) * 100
print(f"Task06b : share of total (%): {pairs_dict(zip(regions, np.round(pct_share, 2)))}")

print(f"Task06c : most populous unit: {regions[np.argmax(census23)]}")
print(f"Task06d : units above 20 million: {regions[np.where(census23 > 20)]}")

ann_growth = 100 * ((census23 / p17) ** (1 / 6) - 1)
print(f"Task06e : annual growth 2017-2023 (% per year): {pairs_dict(zip(regions, np.round(ann_growth, 2)))}")

print(f"Task06f : fastest-growing unit: {regions[np.argmax(ann_growth)]} ({ann_growth.max():.2f}% per year)")

nat_rate = ((census23.sum() / p17.sum()) ** (1 / 6) - 1) * 100
dbl_time = 70 / ann_growth.max()
print(f"Task06 interpret : national growth = {nat_rate:.2f}% per year.")
print(f"Task06 interpret : {regions[np.argmax(ann_growth)]} would double in about {dbl_time:.0f} years at its current rate.")
print(f"Task06 interpret : AJK grew {ann_growth[-1]:.2f}% per year vs national {nat_rate:.2f}% -> "
      f"AJK is {nat_rate - ann_growth[-1]:.2f} points below the national rate (in this sample AJK is essentially flat).")
```

T07:

```python
## Task 07 ##
cities = np.array(["Lahore", "Karachi", "Islamabad", "Quetta", "Peshawar"])
month_names = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                   "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])

temperature = np.array([
    [19, 22, 28, 35, 40, 40, 36, 34, 34, 32, 26, 21],
    [26, 28, 32, 34, 35, 34, 32, 31, 32, 34, 31, 28],
    [17, 19, 24, 30, 36, 38, 35, 33, 33, 30, 24, 19],
    [11, 13, 18, 24, 29, 34, 35, 34, 31, 25, 19, 13],
    [18, 20, 25, 31, 37, 40, 38, 36, 35, 31, 25, 19],
])
precip = np.array([
    [12, 22, 34, 22, 21, 54, 214, 232, 82, 8, 3, 8],
    [8, 10, 9, 4, 1, 6, 40, 25, 13, 1, 2, 5],
    [56, 65, 76, 52, 32, 66, 260, 310, 120, 28, 12, 28],
    [38, 40, 33, 20, 8, 5, 18, 15, 5, 3, 7, 18],
    [33, 48, 63, 48, 23, 21, 86, 95, 43, 23, 20, 29],
])

def pairs_dict(dice_pairs):
    out = {}
    for kar, v in dice_pairs:
        out[str(kar)] = v.item() if hasattr(v, "item") else v
    return out

print(f"Task07a : shape of temp (cities, months): {temperature.shape}")

avg_temp = temperature.mean(axis=1)
print(f"Task07b : yearly mean temperature (deg C): {pairs_dict(zip(cities, np.round(avg_temp, 1)))}")

hottest = month_names[temperature.argmax(axis=1)]
print(f"Task07c : hottest month per city: {pairs_dict(zip(cities, hottest))}")

print(f"Task07d : 5-city monthly average (deg C): {pairs_dict(zip(month_names, np.round(temperature.mean(axis=0), 1)))}")

over35 = (temperature > 35).sum(axis=1)
print(f"Task07e : months above 35 deg C: {pairs_dict(zip(cities, over35))}")

lhr_f = temperature[0] * 1.8 + 32
print(f"Task07f : Lahore monthly temperature (deg F): {np.round(lhr_f, 1)}")

rain_jas = np.sum(precip[:, 6:9], axis=1)
print(f"Task07g : monsoon rain Jul-Sep (mm): {pairs_dict(zip(cities, rain_jas))}")
print(f"Task07g : wettest monsoon city: {cities[np.argmax(rain_jas)]} ({rain_jas.max()} mm)")

pct_share = 100 * rain_jas / precip.sum(axis=1)
print(f"Task07h : monsoon share of annual rain (%): {pairs_dict(zip(cities, np.round(pct_share, 1)))}")

kar = np.flatnonzero(cities == "Karachi")[0]
print(f"Task07 interpret: Karachi has the highest yearly mean ({avg_temp.max():.1f} deg C) "
      f"but {over35[kar]} months above 35 deg C, while cooler-on-average Peshawar has {over35[4]} "
      f"and Lahore {over35[0]}. A high yearly mean does not mean the most extreme heat: "
      f"Karachi is hot all year, the inland cities spike in summer.")
```

T08:

```python
## Task 08 ##
fiscal_years = np.array(["FY2023", "FY2024", "FY2025"])
fy_months = np.array(["Jul", "Aug", "Sep", "Oct", "Nov", "Dec",
                    "Jan", "Feb", "Mar", "Apr", "May", "Jun"])

infl_data = np.array([
    [24.9, 27.3, 23.2, 26.6, 23.8, 24.5, 27.6, 31.5, 35.4, 36.4, 38.0, 29.4],
    [28.3, 27.4, 31.4, 26.9, 29.2, 29.7, 28.3, 23.1, 20.7, 17.3, 11.8, 12.6],
    [4.1, 9.6, 6.9, 7.2, 4.9, 4.1, 2.4, 1.5, 0.7, 0.3, 3.5, 3.2],
])

def pairs_dict(dice_pairs):
    out = {}
    for kar, v in dice_pairs:
        out[str(kar)] = v.item() if hasattr(v, "item") else v
    return out

avg_rate = infl_data.mean(axis=1)
print(f"Task08a: average inflation per year (%): {pairs_dict(zip(fiscal_years, np.round(avg_rate, 2)))}")

fy24 = infl_data[1]
print(f"Task08b: FY2024 months above 25 %: {fy_months[fy24 > 25]} -> {np.sum(fy24 > 25)} months")

delta = infl_data[:, 1:] - infl_data[:, :-1]
print(f"Task08c: month-on-month change (percentage points; negative = drop):\n{np.round(delta, 1)}")

drop_idx = np.argmin(delta, axis=1)
print(f"Task08d: month of biggest drop: {pairs_dict(zip(fiscal_years, fy_months[drop_idx + 1]))} : size (pp): {np.round(delta.min(axis=1), 1)}")

infl_fy25 = infl_data[2]
band = np.where(infl_fy25 < 5, "Low", np.where(infl_fy25 < 10, "Moderate", "High"))
print(f"Task08e: FY2025 labels (Low < 5 %, Moderate 5-10 %, High >= 10 %):\n{np.column_stack((fy_months, infl_fy25, band))}")

print(f"Task08 interpret : average inflation went {avg_rate[0]:.1f}% -> {avg_rate[1]:.1f}% -> {avg_rate[2]:.1f}%. "
      f"Prices did NOT fall; they kept rising, but more and more slowly (disinflation), "
      f"so the price LEVEL is still much higher than at the start of FY2023.")
```

```python
## Task 09 ##
marks = np.array([34, 38, 35, 52, 67, 73, 88, 95, 98, 39,
                  41, 45, 36, 58, 62, 71, 80, 33, 99, 44])
student_rolls = np.arange(1, 21)
CUTOFF = 40

scores_plus5 = np.clip(marks + 5, 0, 100)
print(f"Task09a: original marks: {marks}")
print(f"Task09a: marks after +5 grace (cap 100): {scores_plus5}")

before_pct = np.mean(marks >= CUTOFF) * 100
after = (scores_plus5 >= CUTOFF).mean() * 100
print(f"Task09b: pass % before grace = {before_pct:.1f} %, after grace = {after:.1f} %")

grace_only = np.logical_and(marks < CUTOFF, scores_plus5 >= CUTOFF)
print(f"Task09c: passed only because of grace (roll numbers): {student_rolls[grace_only]}")
print(f"Task09c: their original marks: {marks[grace_only]}")

z_vals = (scores_plus5 - scores_plus5.mean()) / scores_plus5.std()
print(f"Task09d: mean = {scores_plus5.mean():.2f}, std = {scores_plus5.std():.2f}")
print(f"Task09d: z-scores (std devs from mean): {np.round(z_vals, 2)}")
```

```python
## Task 10 ##
digit_img = np.zeros((8, 8), dtype=np.uint8)
digit_img[1, 1:7] = 255
for r, c in zip([2, 3, 4, 5, 6], [6, 5, 5, 4, 4]):
    digit_img[r, c] = 255

binarize = lambda a: (a > 0).astype(int)

print(f"Task10a: the '7' (1 = white):\n{binarize(digit_img)}")

flipped = digit_img[:, ::-1]
print(f"Task10b: flipped left-right:\n{binarize(flipped)}")

negative = 255 - digit_img
print(f"Task10c: inverted (white <-> black):\n{binarize(negative)}")

img_crop = digit_img[:7, 1:]
print(f"Task10d: cropped 7x7 region, shape: {img_crop.shape}\n{binarize(img_crop)}")

brighter = np.clip(digit_img.astype(np.int16) + 60, 0, 255).astype(np.uint8)
print(f"Task10e: brightened (+60, capped at 255); unique values: {np.unique(brighter)}")

print(f"Task10f: shape = {digit_img.shape} : dtype = {digit_img.dtype} : white pixels = {np.count_nonzero(digit_img == 255)}")
```
