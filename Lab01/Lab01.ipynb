```python
def get_domain(email):
    at_index = email.index("@")
    return email[at_index + 1:]

result = get_domain("user@domain.com")
```

```python
print("Q7:", result)
```

```python
def count_dog(text):
    small = text.lower()
    return small.count("dog")

result = count_dog("Dog dog DOG cat")
```

Task 10

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed = speed - 5
    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

result1 = caught_speeding(60, False)
result2 = caught_speeding(65, False)
result3 = caught_speeding(90, False)
```

```python
print(f"Q10: {result1}, {result2}, {result3}")
```
